```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
import urllib.robotparser
rp = urllib.robotparser.RobotFileParser()
rp.set_url("https://dnd5e.wikidot.com//robots.txt")
rp.read()
print(rp.can_fetch("*", "https://dnd5e.wikidot.com/"))
```

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/81.0.4044.138 Safari/537.36'
}
```

```python
r = requests.get('https://dnd5e.wikidot.com/', headers=headers)
```

```python
soup = BeautifulSoup(r.content, 'lxml')
```

```python
soup
```

```python
page_contents = soup.find_all('a')[44:]
page_contents = page_contents[:701]
print(page_contents)
```

```python
specific_tag = None
for tag in page_contents:
    if tag.get_text().strip() == 'All Spells':
        specific_tag = tag
        break

print(page_contents.index(specific_tag))
print(len(page_contents))
print(page_contents[0])
```

```python
for tag in soup:
    print(tag)
    # You can also inspect the parent element
    # print(tag.parent)
    # Or previous siblings
    # print(tag.previous_sibling)
    print("-" * 20) # Separator for clarity
```

```python
subtitles = []
subcategories = []

for tag in page_contents:
    # Check if the parent of the <a> tag is an <li> element
    if tag.parent and tag.parent.name == 'li':
        subcategories.append(tag)
    else:
        subtitles.append(tag)

print("Subtitles found:", [tag.get_text() for tag in subtitles])
print("\nSubcategories found:", [tag.get_text() for tag in subcategories])
```
